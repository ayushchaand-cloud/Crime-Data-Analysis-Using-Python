# Crime Dataset Analysis

**Dataset:** crime_dataset_india.csv

This notebook contains basic data analysis and visualization using Pandas, Matplotlib and Seaborn.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

f = pd.read_csv("crime_dataset_india.csv")

f.head()
```

## 1. Basic information

```python
f.shape
```

```python
f.info()
```

```python
f.describe()
```

```python
f.isnull().sum()
```

```python
f.duplicated().sum()
```

## 2. Crime wise analysis

```python
crime_count = f['Crime Description'].value_counts()
crime_count
```

```python
plt.figure(figsize=(12,6))
crime_count.head(10).plot(kind='bar')
plt.title('Top 10 Crime Types')
plt.xlabel('Crime')
plt.ylabel('Number of Cases')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 3. City wise cases

```python
city_count = f['City'].value_counts()
city_count.head(10)
```

```python
plt.figure(figsize=(10,6))
city_count.head(10).plot(kind='barh')
plt.title('Top 10 Cities by Number of Cases')
plt.xlabel('Number of Cases')
plt.ylabel('City')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## 4. Victim gender

```python
gender_count = f['Victim Gender'].value_counts()
gender_count
```

```python
plt.figure(figsize=(7,7))
plt.pie(gender_count.values, labels=gender_count.index, autopct='%1.1f%%')
plt.title('Victim Gender Distribution')
plt.show()
```

## 5. Victim age

```python
plt.figure(figsize=(9,5))
sns.histplot(f['Victim Age'], bins=15, kde=True)
plt.title('Distribution of Victim Age')
plt.xlabel('Victim Age')
plt.ylabel('Number of Cases')
plt.show()
```

## 6. Crime domain

```python
domain_count = f['Crime Domain'].value_counts()
domain_count
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=f, x='Crime Domain')
plt.title('Cases by Crime Domain')
plt.xlabel('Crime Domain')
plt.ylabel('Number of Cases')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

## 7. Case status

```python
case_status = f['Case Closed'].value_counts()
case_status
```

```python
plt.figure(figsize=(7,5))
case_status.plot(kind='bar')
plt.title('Case Closed vs Open Cases')
plt.xlabel('Case Status')
plt.ylabel('Number of Cases')
plt.xticks(rotation=0)
plt.show()
```

## 8. Crime trend by year

```python
f['Date of Occurrence'] = pd.to_datetime(
    f['Date of Occurrence'],
    dayfirst=True,
    errors='coerce'
)

f['Year'] = f['Date of Occurrence'].dt.year

year_count = f['Year'].value_counts().sort_index()
year_count
```

```python
plt.figure(figsize=(9,5))
plt.plot(year_count.index, year_count.values, marker='o')
plt.title('Crime Cases by Year')
plt.xlabel('Year')
plt.ylabel('Number of Cases')
plt.grid()
plt.show()
```

## 9. Most common weapons

```python
weapon_count = f['Weapon Used'].value_counts()
weapon_count
```

```python
plt.figure(figsize=(9,5))
sns.barplot(x=weapon_count.head(6).values, y=weapon_count.head(6).index)
plt.title('Most Common Weapons Used')
plt.xlabel('Number of Cases')
plt.ylabel('Weapon')
plt.tight_layout()
plt.show()
```

## 10. Average police deployment

```python
police_by_crime = (
    f.groupby('Crime Description')['Police Deployed']
    .mean()
    .sort_values(ascending=False)
)

police_by_crime.head(10)
```

```python
plt.figure(figsize=(10,6))
police_by_crime.head(10).plot(kind='barh')
plt.title('Average Police Deployed for Top Crime Types')
plt.xlabel('Average Police Deployed')
plt.ylabel('Crime')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## 11. Crime and victim gender

```python
gender_crime = pd.crosstab(f['Crime Description'], f['Victim Gender'])
gender_crime.head()
```

```python
top_crimes = f['Crime Description'].value_counts().head(8).index
gender_crime_top = gender_crime.loc[top_crimes]

gender_crime_top.plot(kind='bar', figsize=(12,6))
plt.title('Top Crimes by Victim Gender')
plt.xlabel('Crime')
plt.ylabel('Number of Cases')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 12. Female cybercrime cases by city

```python
cyber_female_city = (
    f[(f['Crime Description'] == 'CYBERCRIME') &
      (f['Victim Gender'] == 'F')]['City']
    .value_counts()
)

cyber_female_city.head(10)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(
    range(len(cyber_female_city.head(10))),
    cyber_female_city.head(10).values,
    s=80
)
plt.xticks(range(len(cyber_female_city.head(10))), cyber_female_city.head(10).index, rotation=45)
plt.title('Female Cybercrime Cases in Top Cities')
plt.xlabel('City')
plt.ylabel('Number of Cases')
plt.tight_layout()
plt.show()
```

## 13. Female kidnapping cases by city

```python
kidnap_female_city = (
    f[(f['Crime Description'] == 'KIDNAPPING') &
      (f['Victim Gender'] == 'F')]['City']
    .value_counts()
)

kidnap_female_city.head(10)
```

```python
plt.figure(figsize=(10,6))
plt.pie(
    kidnap_female_city.head(6).values,
    labels=kidnap_female_city.head(6).index,
    autopct='%1.1f%%'
)
plt.title('Female Kidnapping Cases by City')
plt.show()
```

## 14. Female identity theft and age

```python
id_theft_female = f[
    (f['Crime Description'] == 'IDENTITY THEFT') &
    (f['Victim Gender'] == 'F')
]

id_theft_female[['City', 'Victim Age']].head()
```

```python
top_id_cities = id_theft_female['City'].value_counts().head(6).index
plot_data = id_theft_female[id_theft_female['City'].isin(top_id_cities)]

plt.figure(figsize=(11,6))
sns.boxplot(data=plot_data, x='City', y='Victim Age')
plt.title('Age of Female Identity Theft Victims')
plt.xlabel('City')
plt.ylabel('Victim Age')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 15. Simple observations

```python
print("Total cases:", len(f))
print("Most common crime:", f['Crime Description'].value_counts().idxmax())
print("City with most cases:", f['City'].value_counts().idxmax())
print("Most common victim gender:", f['Victim Gender'].value_counts().idxmax())
print("Most common weapon:", f['Weapon Used'].value_counts().idxmax())
print("Most common crime domain:", f['Crime Domain'].value_counts().idxmax())
```
